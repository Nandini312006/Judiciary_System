```python
import requests
import json
import ollama
```

```python
def summarize_legal_case_ollama(case_text: str, model_name: str = "llama3.2") -> str:
    """
    Summarizes a legal case using local Ollama model (free, no API key)
    """
    prompt = f"""You are an AI legal assistant.

Your task is to summarize the given legal case.

Output format:

CASE SUMMARY:

* Facts:
* Issues:
* Judgment:

Rules:
* Keep it short and clear
* Do not add extra information
* Do not assume missing details
* If judgment is not present, write "Not available"

Case:
{case_text}"""

    # Call Ollama's local API using requests (no ollama library needed)
    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": model_name,
            "prompt": prompt,
            "stream": False,
            "temperature": 0.0
        }
    )
    
    result = response.json()
    return result["response"].strip()
```

```python
if __name__ == "__main__":
    sample_case = """
    In the case of Smith v. Jones (2023), the plaintiff argued breach of contract.
    The defendant claimed the contract was void for lack of consideration.
    The court found that consideration existed and ruled for the plaintiff.
    """
    
    print("Summarizing legal case using local Ollama...")
    summary = summarize_legal_case_ollama(sample_case)
    print(summary)
```
